#### 95% Confidence Interval for a Sample Mean (t-distribution)
--------------------------------------------------------------------------------
Scenario:

A small shop records daily ice cream sales over 7 days and wants to estimate the
average daily sales for the whole month, based only on this sample.

```python
import numpy as np
from scipy import stats

ice_cream_sales = [50, 52, 48, 47, 51, 53, 49]

# Step 1: Mean and standard error
mean_sales = np.mean(ice_cream_sales)
std_dev = np.std(ice_cream_sales, ddof=1)
n = len(ice_cream_sales)
std_error = std_dev / np.sqrt(n)

# Step 2: t-score for a 95% confidence level
confidence = 0.95
t_score = stats.t.ppf((1 + confidence) / 2, df=n - 1)

# Step 3: Margin of error
margin_of_error = t_score * std_error

# Step 4: Confidence interval
lower_bound = mean_sales - margin_of_error
upper_bound = mean_sales + margin_of_error

print(f"Average sales: {mean_sales:.2f}")
print(f"95% confidence interval: ({lower_bound:.2f}, {upper_bound:.2f})")
```
