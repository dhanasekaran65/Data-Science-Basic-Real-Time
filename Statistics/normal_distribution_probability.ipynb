#### Probability Calculation for a Continuous Random Variable (Normal Distribution)
Problem:

The length of a human pregnancy is normally distributed with a mean of 272 days
and a standard deviation of 9 days.

    1. State the random variable.
    2. Find the probability of a pregnancy lasting more than 280 days.
    3. Find the probability of a pregnancy lasting less than 250 days.
    4. Find the probability that a pregnancy lasts between 265 and 280 days.

```python
from scipy import stats
MEAN = 272
STD_DEV = 9
# 1. Random Variable (RV): Length of a human pregnancy, in days.
print("Random Variable: Length of a human pregnancy (in days)\n")

# 2. P(pregnancy > 280 days)
p_more_than_280 = 1 - stats.norm.cdf(x=280, loc=MEAN, scale=STD_DEV)
print(f"P(X > 280) = {p_more_than_280:.4f}")

# 3. P(pregnancy < 250 days)
p_less_than_250 = stats.norm.cdf(x=250, loc=MEAN, scale=STD_DEV)
print(f"P(X < 250) = {p_less_than_250:.4f}")

# 4. P(265 < pregnancy < 280 days)
p_between_265_and_280 = stats.norm.cdf(x=280, loc=MEAN, scale=STD_DEV) - stats.norm.cdf(
    x=265, loc=MEAN, scale=STD_DEV
)
print(f"P(265 < X < 280) = {p_between_265_and_280:.4f}")
```
