## Perform Correlation Analysis
- Correlation Analysis helps you understand the relationship between two or more variables in a dataset.

what is Correlation ?
- Correlation measures how strongly two variables are related.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Study_Hours':[2,4,6,8,10],
    'Marks':[40,50,60,80,95],
    'Sleep_Hours':[8,7,6,5,4]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
corr = df.corr()
print(corr)
```

```python
plt.figure(figsize=(7,5))
sns.heatmap(corr,annot=True,cmap='coolwarm',linewidths=0.5,linecolor='black',fmt='.2f')
plt.title("Correlation Student Marks")
plt.show()
```

#### Identify Strong Correlation

```python
corr=df.corr()
print(corr['Marks'].sort_values(ascending=False))
```

```python
corr = df.corr()
print(corr['Study_Hours'].sort_values(ascending=False))
```

### Plot Scatter Plot for Validation

Why?
- Correlation gives a number, but scatter plots show the actual relationship visually.

```python
sns.scatterplot(df,x='Study_Hours',y='Marks')
plt.title("Study Hours Vs Marks")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

mt_cars = pd.read_csv("mtcars.csv")

corr_matrix = mt_cars.select_dtypes(include='number').corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix,
           annot=True,
           cmap='viridis',
           linewidths=1,
           linecolor='red',
           fmt='.2f')
plt.title("Correlation of Mtcars")
plt.show()
```

### Find Highly Correlated Feature

```python
corr_matrix = mt_cars.corr(numeric_only=True)
high_corr = corr_matrix[
    (corr_matrix > 0.7) &
    (corr_matrix < 1.0)
]
print(high_corr)
```
