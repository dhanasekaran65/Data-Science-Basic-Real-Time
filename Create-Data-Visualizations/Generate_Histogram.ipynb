```python
import matplotlib.pyplot as plt
student_marks = {
    "maths": [55, 23, 67, 45, 89, 92, 35, 50, 20],
    "science": [65, 34, 78, 64, 50, 35, 90, 15, 20],
    "english": [78, 35, 65, 98, 29, 45, 60, 30, 40, 50]
}
grade = [0,20,35,50,75,100]
fig, ax = plt.subplots()
ax.hist(student_marks['maths'],grade,
        fc='g',
        ec='black')
ax.set( yticks =[1,2,3,4])
ax.set_title("Maths Marks Histogram")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
student_marks = {
    "maths": [55, 23, 67, 45, 89, 92, 35, 50, 20],
    "science": [65, 34, 78, 64, 50, 35, 90, 15, 20],
    "english": [78, 35, 65, 98, 29, 45, 60, 30, 40, 50]
}
fig, ax = plt.subplots()
ax.hist(student_marks['maths'],
        bins=6,
        fc='g',
        ec='black')
ax.set( yticks =[1,2,3,4])
ax.set_title("Maths Marks Histogram")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")
plt.show()
```

```python
student_marks = {
    "maths": [55, 23, 67, 45, 89, 92, 35, 50, 20],
    "science": [65, 34, 78, 64, 50, 35, 90, 15, 20],
    "english": [78, 35, 65, 98, 29, 45, 60, 30, 40, 50]
}
fig, ax = plt.subplots()
ax.hist(student_marks['maths'], bins=5, alpha=1,ec='red', label='Maths')
ax.hist(student_marks['science'], bins=5, alpha=1,ec='red', label='Science')
ax.hist(student_marks['english'], bins=5, alpha=1,ec='red', label='English')
ax.set(yticks=[1,2,3,4,5])
ax.set_title("Student Marks Distribution")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")
ax.legend()

plt.show()
```
