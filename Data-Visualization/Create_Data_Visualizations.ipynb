```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("employee_analytics.csv")
print(df)
```

```python
df.head()
```

#### 1. Bar Chart

Compare average salary by department.

```python
plt.figure(figsize=(10,5))
df.groupby('Department')['Salary'].mean().plot(kind='bar')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Salary')
plt.show()
```

#### 2. Count Plot

Count employees in each department.

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Department', data=df)
plt.title('Employees per Department')
plt.show()
```

#### 3. Pie Chart

Gender distribution.

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%'
)
plt.title('Gender Distribution')
plt.show()
```

#### 4. Histogram

Age distribution.

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'],bins=5)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

#### KDE Plot

Salary density.

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Salary'], fill=True)
plt.title('Salary Density')
plt.show()
```

#### 6. Box Plot

Detect salary outliers.

```python
plt.figure(figsize=(8,5))
sns.boxplot(y='Salary', data=df)
plt.title('Salary Box Plot')
plt.show()
```

#### 7. Violin Plot

Salary distribution by gender.

```python
plt.figure(figsize=(8,5))
sns.violinplot(
    x='Gender',
    y='Salary',
    data=df
)
plt.show()
```

#### 8. Scatter Plot

Experience vs Salary.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Experience_Years',
    y='Salary',
    data=df
)
plt.show()
```

#### 9. Line Chart

Salary trend across employees.

```python
plt.figure(figsize=(12,5))
plt.plot(df['Employee_ID'], df['Salary'])
plt.title('Salary Trend')
plt.show()
```

#### 10. Pair Plot

Relationships between numeric variables.

```python
sns.pairplot(
    df[['Age',
        'Salary',
        'Experience_Years',
        'Performance_Rating']]
)
plt.show()
```

#### 11. Heatmap

Correlation matrix.(Correlation)

```python
plt.figure(figsize=(8,6))
corr = df.select_dtypes(include='number').corr()
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()
```

#### 12. Area Chart

Age and Experience comparison.

```python
df[['Age','Experience_Years']].head(50).plot.area(figsize=(10,5))
plt.show()
```

#### 13. Stacked Bar Chart

Department vs Attrition.

```python
pd.crosstab(
    df['Department'],
    df['Attrition']
).plot(
    kind='bar',
    stacked=True,
    figsize=(10,5)
)
plt.title('Department vs Attrition')
plt.show()
```

#### 14. Donut Chart

Department distribution.

```python
dept = df['Department'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(
    dept,
    labels=dept.index,
    autopct='%1.1f%%'
)
centre = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre)
plt.title('Department Distribution')
plt.show()
```

#### 15. Swarm Plot

Salary by department.

```python
plt.figure(figsize=(12,5))
sample_df = df.sample(300)
sns.swarmplot(
    x='Department',
    y='Salary',
    data=sample_df
)
plt.show()
```

#### 16. Strip Plot

Salary spread by department.

```python
plt.figure(figsize=(12,5))
sns.stripplot(
    x='Department',
    y='Salary',
    data=df,
    jitter=True
)
plt.show()
```

#### 17. Joint Plot

Experience vs Salary.

```python
sns.jointplot(
    x='Experience_Years',
    y='Salary',
    data=df,
    kind='scatter'
)
plt.show()
```

#### 18. Regression Plot

Trend line.

```python
plt.figure(figsize=(8,5))
sns.regplot(
    x='Experience_Years',
    y='Salary',
    data=df
)
plt.title('Experience vs Salary')
plt.show()
```

#### 19. Cluster Map

Clustered correlation matrix.

```python
corr = df.select_dtypes(include='number').corr()
sns.clustermap(
    corr,
    annot=True
)
plt.show()
```

#### 20. Hexbin Plot

Dense scatter visualization.

```python
df.plot.hexbin(
    x='Experience_Years',
    y='Salary',
    gridsize=20,
    figsize=(8,5)
)
plt.title('Hexbin Plot')
plt.show()
```
