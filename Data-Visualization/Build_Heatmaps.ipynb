## 1.Build Heatmaps in Python

- A Heatmap is a graphical representation of data where values are represented by colors. 
- It helps identify patterns, relationships, and correlations quickly.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sample_data = np.array([
    [20,40,60,80],
    [10,30,50,70],
    [90,100,110,120]
])
```

```python
sns.heatmap(sample_data)
plt.title("Basic Heatmap")
plt.show()
```

#### Concept
- Dark colors = Lower values
- Bright colors = Higher values

### 1.1  Add Annotation

```python
sns.heatmap(sample_data,annot=True)
plt.title("Heatmap with values")
plt.show()
```

### 1.2 Change Color

using Coolwarm

```python
sns.heatmap(sample_data,annot=True,cmap = 'coolwarm')
plt.title("Heatmap using Coolwarm")
plt.show()
```

###### 
using viridis

```python
sns.heatmap(sample_data,annot=True,cmap = 'viridis')
plt.title("Heatmap using Viridis")
plt.show()
```

```python
sns.heatmap(sample_data,annot=True,cmap = 'YlGnBu')
plt.title("Heatmap using YlGnBu")
plt.show()
```

### 1.3 Heatmap using Pandas DataFrame

```python
df = pd.DataFrame({
    "Maths":[50,35,65,70,89,95],
    "Science":[45,35,56,76,89,99],
    "Social Science":[50,69,70,90,96,80]
})
print(df)
```

```python
sns.heatmap(df,annot = True,cmap = 'Reds')
plt.title("Student Marks")
plt.show()
```

### 1.4 Correlation Heatmap

```python
df = pd.DataFrame({
    "Age":[21,23,30,43,27,29],
    "Salary":[20000,25000,50000,60000,35000,40000],
    "Experience":[1,2,5,8,3,4]
})
print(df)
```

```python
corr_matrix = df.corr()
print(corr_matrix)
```

```python
sns.heatmap(corr_matrix,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

#### Understand correlation values
 Value -- Meaning
- 1.0  -- Strong Positive
- 0.8  -- High Positive
- 0.5  -- Moderate Positive
- 0    -- No Relationship
- -0.5 -- Moderate Negative
- -1.0 -- Strong Negative


### 1.5 professional heatmap format

```python
df = pd.DataFrame({
    "Age":[21,23,30,43,27,29],
    "Salary":[20000,25000,50000,60000,35000,40000],
    "Experience":[1,2,5,8,3,4]
})
corr_matrix = df.corr()
plt.figure(figsize = (8,6))
sns.heatmap(corr_matrix,
            annot=True,
            cmap="viridis",
           linewidths=1,
           linecolor = "black",
           fmt='.2f'
)
plt.title("Professional Format Correlation Heatmap")
plt.show()
```

#### Parameter - Purpose
- figsize -- Chart size
- linewidths -- Cell border width
- linecolor -- Border color
- fmt='.2f' -- Two decimal places


```python
employee_data = pd.DataFrame({
    'Age':[25,30,35,40,45],
    'Salary':[30000,40000,50000,60000,70000],
    'Experience':[1,3,5,7,9],
    'Performance':[2,3,4,4,5]
})
corr = employee_data.corr()
plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    annot=True,
    cmap='YlGnBu'
)
plt.title("Employee Analysis Heatmap")
plt.show()
```

## Build Heatmap

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
mt_cars = pd.read_csv("mtcars.csv")
corr = mt_cars.corr(numeric_only = True)
plt.figure(figsize=(20,8))
sns.heatmap(corr,annot=True,cmap = 'viridis',linewidths=1,linecolor='black',fmt = '.2f')
plt.title("Feature Correlation Heatmap")

plt.show()
```
