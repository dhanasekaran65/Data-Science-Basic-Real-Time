## 1. Analyze Trends in Datasets 

- Trend analysis helps identify patterns, increases, decreases, and changes in data over time. 
- It is commonly used in sales, finance, marketing, HR, and business analytics.

#### 1.1 Import Necessary Libraries 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### 1.2 Load Dataset

```python
sales_data = pd.read_csv("sales_data.csv")
print(sales_data)
```

```python
print(sales_data.head())
```

#### 1.3 Basic Statistics

```python
print(sales_data.describe())
```

#### 1.4 Visualize Trend Using Line Chart

```python
plt.figure(figsize=(6,4))
plt.plot(sales_data['Month'],
        sales_data['Sales'],
        marker='o')
plt.title("Visualize Line Chart")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

#### Interpretation
- Upward line → Growth trend
- Downward line → Declining trend
- Flat line → Stable trend

#### 1.5 Calculate Growth Rate
Formula:

Growth Rate = Previous Value - Current Value  / Previous Value  x 100

```python
sales_data['Growth_Rate'] = sales_data['Sales'].pct_change() * 100
print(sales_data)
```

#### 1.6 Moving Average Trend

Moving Average removes short-term fluctuations.

```python
sales_data['Moving_Avg'] = sales_data['Sales'].rolling(window=3).mean()
print(sales_data)
```

#### 1.7 Plot Moving Average

```python
plt.figure(figsize=(10,5))
plt.plot(sales_data['Month'],sales_data['Sales'],label='Actual Sales')

plt.plot(sales_data['Sales'],
        sales_data['Moving_Avg'],
        label='3-Month Moving Average')
plt.legend()
plt.title("Sales Trends Analysis")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(sales_data['Month'],sales_data['Sales'],label = 'Actual Sales')
plt.plot(sales_data['Month'],sales_data['Growth_Rate'],label = 'Sales Growth')
plt.legend()
plt.title('Growth Sales Trends Analysis')
plt.show()
         
```

#### 1.8 Detect Highest and Lowest Sales

```python
print("Highest Sales:")
print(sales_data.loc[sales_data['Sales'].idxmax()])

print("Lowest Sales:")
print(sales_data.loc[sales_data['Sales'].idxmin()])
```

#### 1.9 Year-over-Year or Month-over-Month Comparison

```python
sales_data['Different'] = sales_data['Sales'].diff()
print(sales_data[['Month','Sales','Different']])
```
