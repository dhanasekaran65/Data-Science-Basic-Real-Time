###### 10. Correlation Analysis

###### 13. Outlier Detection

```python
sns.boxplot(data=df)
plt.show()
```

###### using IQR Method

```python
Q1 = df['Revenue'].quantile(0.25)
Q3 = df['Revenue'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1-1.5*IQR
upper = Q3 + 1.5 * IQR
outliers = df[
    (df['Revenue'] < lower) |
    (df['Revenue'] > upper)
]
print(outliers)
    
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("datasets.csv")
# Basic Information

print(df.head())
print(df.info())
print(df.describe())
# Missing Values

print(df.isnull().sum())
# Duplicates

print(df.duplicated().sum())
# Correlation

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')
plt.show()
# Histogram

df.hist(figsize=(10,8))
plt.show()
# Box Plot

sns.boxplot(data=df)
plt.show()
# Scatter Plot

sns.scatterplot(x='Employees',
                y='Revenue',
                data=df)
plt.show()

# Pair Plot

sns.pairplot(df)
plt.show()
```
